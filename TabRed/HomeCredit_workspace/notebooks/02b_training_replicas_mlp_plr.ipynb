# 02b — Training Replicas (MLP-PLR)

Trains R MLP-PLR replicas per window pair (A, B) on **Colab GPU**, following the same
structure and output contract as notebook 02. MLP-PLR uses **Periodic → Linear → ReLU**
embeddings per numeric feature (Gorishniy et al. 2022, *On Embeddings for Numerical
Features in Tabular Deep Learning*) followed by an MLP backbone on the concatenation of
embedded numerics + raw binaries + one-hot dummies (the same one-hot expansion notebook 02 fits per (window, side)).

The output schema (`predictions.npz` keys, `reference_*.joblib`, run-params) matches
notebook 02 exactly, so notebook 04 consumes MLP-PLR results the same way it consumes
XGBoost / LR results.

## Per-window preprocessing

Three preprocessing transforms are involved. Their fit scope differs by purpose:

- **OrdinalEncoder** (categorical vocabulary): a single `OrdinalEncoder(min_frequency=1/100, handle_unknown='use_encoded_value', unknown_value=-1)` is fit **once per (window, side)** on that window's full training rows and saved to `replicas_<side>/encoder.joblib`. Every bootstrap replica of that side reuses the same encoder so categorical integer codes are stable across replicas — a prerequisite for SHAP / LIME / drift comparisons in 03 / 03b / 04. The same shared encoder is also used inside Optuna's TimeSeriesSplit folds.

- **OneHotEncoder** (`one_hot`): fit **once per (window, side)** on the OrdinalEncoder's integer output from the full window training rows and saved to `replicas_<side>/one_hot.joblib` — the same artefact notebook 02 fits and saves. Reused by all bootstrap replicas of that side and across all CV folds during tuning. Encoding nominal categories as one-hot dummies avoids imposing a fake monotonic ordering through integer codes. Rare-category collapsing happens upstream in the OrdinalEncoder (`min_frequency=1/100`); unseen categories at inference get all-zero dummies via `handle_unknown='ignore'`.

- **Numeric StandardScaler** (`scaler` — per-replica): fit **per bootstrap replica** on that replica's numeric training slice, bundled inside `model_r{r}.joblib`. Numeric distributions can differ across bootstrap samples, so this scaler intentionally varies — it is part of the per-replica preprocessing pipeline that the model is trained on.

Binary indicators stay raw 0/1 (no scaling).

## Class weighting (operational retraining pipeline)

MLP-PLR is fitted with a **weighted binary cross-entropy** loss using balanced sample weights equivalent to sklearn's `class_weight='balanced'`:

```
w_pos = n / (2 * n_pos)        w_neg = n / (2 * n_neg)        n = n_pos + n_neg
```

This applies to every fit performed in the notebook — Optuna folds and final replica training — and is part of the operational retraining pipeline rather than a per-dataset toggle. Concretely:

- The side/window-level `{w_pos, w_neg}` mapping is computed **once per side per pair** from the bootstrap source (the training core, before the chronological early-stopping tail) and reused by every replica of that side. The bootstrap remains class-stratified.
- Optuna folds compute their own balanced weights inside `train_one_mlp_plr` from the actual optimization-training portion (post internal ES-tail split), so the internal early-stopping tail is never used to derive weights.
- A and B may end up with **different** weights because their class balance can differ across temporal windows; this is intentional and is part of what we measure as end-to-end explanation drift.
- The per-row weighted BCE loss is averaged across the batch. Under the balanced scheme the average sample weight is 1, which keeps the loss scale comparable to unweighted training and avoids destabilising AdamW's effective step size. With HomeCredit's ~3% positive rate, w_pos ≈ 16 and w_neg ≈ 0.5 — the loss is no longer dominated by the (correctly-classified) majority, which was the failure mode flagged in the previous version's notes.
- Weights are **never clipped**. Validation PR-AUC, the early-stopping signal, and evaluation PR-AUC / ROC-AUC remain **unweighted** — `average_precision_score` and `roc_auc_score` are called without `sample_weight`.

The realised `{n_pos, n_neg, w_pos, w_neg}` for each side / window / pair is written to `class_weights.json` inside each pair directory; the scheme itself participates in the `run_params.json` staleness fingerprint, so any change to the weighting policy invalidates cached predictions and downstream SHAP / LIME / drift artefacts.

## Forward path

```
[ PLR(X_num_scaled_per_replica), X_bin_raw, X_dummies_per_side ]  →  MLP backbone  →  pre-sigmoid logit
```

PLR (Periodic → Linear → ReLU) is applied **only** to the true numerics. Binary indicators and one-hot dummies both bypass PLR and enter the MLP backbone unchanged. Notebooks 03 and 03b construct the same input layout when calling SHAP / LIME on each replica.

## GPU reproducibility caveat

Training on GPU is not bit-reproducible even with fixed seeds — cuDNN's non-deterministic
algorithms introduce small variance that persists under `cudnn.deterministic=True`. This
low-level noise is part of the real replica-to-replica variability we're measuring.

## Seeding convention (shared with notebook 02)

For replica `r` of pair `pid`, with `SEED_BASE = 42`:
- A bootstrap seed: `SEED_BASE + pid*10_000 + r*2`
- A model seed:     `SEED_BASE + pid*10_000 + r*2 + 1`
- B bootstrap seed: `SEED_BASE + pid*10_000 + 5_000 + r*2`
- B model seed:     `SEED_BASE + pid*10_000 + 5_000 + r*2 + 1`

## HomeCredit-specific notes

- **Heavy class imbalance.** Default rate is ~3.2% (vs Homesite's ~19%). Unweighted PR-AUC values will be much lower in absolute terms — typically 0.05–0.15 — and ROC-AUC is the more comparable metric across implementations. The class-weighting policy above is what compensates for the imbalance during fitting; PR-AUC reporting itself remains on the natural distribution so values stay comparable to other Home Credit baselines.
- **Larger feature space.** After per-window one-hot expansion, the combined matrix has ~1100–1500 columns (HomeCredit's `n_num + n_bin ≈ 720` plus 400–800 dummies depending on category counts after the `min_frequency=0.01` threshold). PLR embedding dimensions are unaffected — they are tuned per window — but the MLP backbone's input dimension grows. End-to-end runtime on a single Colab T4 GPU lands around 6–9 hours across all 4 pairs × both sides; an L4 or A100 is correspondingly faster.

## Model definitions

Note: keep these classes byte-identical with notebook 03's copy (used during SHAP/explanation extraction).

````python
class PLREmbedding(nn.Module):
    """Vectorised PLR embedding for all numeric features at once.

    Per feature (no weight sharing across features):
        periodic(x)  :  v = 2π · c · x        c ∈ ℝ^k, c ~ N(0, sigma_init^2)
                        [sin(v), cos(v)]      → ℝ^(2k)
        linear       :  W·[sin,cos] + b       W ∈ ℝ^(2k × d), b ∈ ℝ^d
        relu         :  max(0, ·)             → ℝ^d
    """

    def __init__(self, n_num_features: int, k_periodic: int,
                 sigma_init: float, d_embedding: int):
        super().__init__()
        self.n_num_features = n_num_features
        self.k_periodic     = k_periodic
        self.d_embedding    = d_embedding

        # Per-feature coefficient vectors c ~ N(0, sigma_init^2)
        self.c = nn.Parameter(torch.randn(n_num_features, k_periodic) * sigma_init)

        # Per-feature linear weights & biases (no weight sharing)
        self.W = nn.Parameter(torch.empty(n_num_features, 2 * k_periodic, d_embedding))
        self.b = nn.Parameter(torch.zeros(n_num_features, d_embedding))
        nn.init.kaiming_uniform_(self.W, a=math.sqrt(5))

    def forward(self, x_num: torch.Tensor) -> torch.Tensor:
        # x_num: (batch, n_num_features)
        v  = (2.0 * math.pi) * self.c.unsqueeze(0) * x_num.unsqueeze(-1)
        pe = torch.cat([torch.sin(v), torch.cos(v)], dim=-1)
        out = torch.einsum('bnk,nkd->bnd', pe, self.W) + self.b.unsqueeze(0)
        out = torch.relu(out)
        return out.reshape(out.shape[0], -1)


class MLPPLR(nn.Module):
    """MLP-PLR with three input groups: numerics (PLR-embedded), binaries (raw 0/1),
    and one-hot dummies (raw 0/1, bypass PLR).

    Outputs a single logit per instance (no sigmoid — use BCEWithLogitsLoss).
    Column routing is stored as non-trainable buffers so it travels with state_dict.
    """

    def __init__(self, n_num_features: int, n_bin_features: int, n_dummies: int,
                 k_periodic: int, sigma_init: float, d_embedding: int,
                 n_layers: int, hidden_dim: int, dropout: float,
                 num_col_idx, bin_col_idx, dummy_col_idx):
        super().__init__()
        self.n_num_features = n_num_features
        self.n_bin_features = n_bin_features
        self.n_dummies      = n_dummies

        # Column routing as buffers (survive state_dict save/load).
        self.register_buffer('num_col_idx_buf',   torch.as_tensor(num_col_idx,   dtype=torch.long))
        self.register_buffer('bin_col_idx_buf',   torch.as_tensor(bin_col_idx,   dtype=torch.long))
        self.register_buffer('dummy_col_idx_buf', torch.as_tensor(dummy_col_idx, dtype=torch.long))

        self.plr = PLREmbedding(n_num_features, k_periodic, sigma_init, d_embedding)

        # MLP backbone — input combines: PLR-embedded numerics + raw binaries + one-hot dummies.
        input_dim = n_num_features * d_embedding + n_bin_features + n_dummies
        layers = []
        in_dim = input_dim
        for _ in range(n_layers):
            layers.extend([
                nn.Linear(in_dim, hidden_dim),
                nn.ReLU(),
                nn.Dropout(dropout),
            ])
            in_dim = hidden_dim
        self.backbone = nn.Sequential(*layers)
        self.head     = nn.Linear(hidden_dim, 1)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # x: (batch, n_num + n_bin + n_dummies) — combined matrix [X | one-hot dummies].
        x_num     = x.index_select(1, self.num_col_idx_buf)     # (batch, n_num)
        x_bin     = x.index_select(1, self.bin_col_idx_buf)     # (batch, n_bin) — raw 0/1
        x_dummies = x.index_select(1, self.dummy_col_idx_buf)   # (batch, n_dummies) — raw 0/1
        emb       = self.plr(x_num)                              # (batch, n_num * d_emb)
        z         = torch.cat([emb, x_bin, x_dummies], dim=-1)   # (batch, input_dim)
        h         = self.backbone(z)                             # (batch, hidden_dim)
        return self.head(h).squeeze(-1)                          # (batch,)


print('Model classes defined.')
````

## Helper functions

`stratified_bootstrap` and `compute_flagged_topk` are byte-identical with notebook 02 so cross-model-type replica alignment holds.

````python
def stratified_bootstrap(idx: np.ndarray, Y: np.ndarray, seed: int) -> np.ndarray:
    """Stratified bootstrap: sample with replacement, preserving class ratio."""
    rng = np.random.default_rng(seed)
    pos = idx[Y[idx] == 1]
    neg = idx[Y[idx] == 0]
    boot_pos = rng.choice(pos, size=len(pos), replace=True)
    boot_neg = rng.choice(neg, size=len(neg), replace=True)
    out = np.concatenate([boot_pos, boot_neg])
    rng.shuffle(out)
    return out


def compute_flagged_topk(p_hat_A, p_hat_B, k_frac):
    """Top-K% flagged set ranked by max(p_hat_A, p_hat_B)."""
    assert p_hat_A.shape == p_hat_B.shape
    n = len(p_hat_A)
    k = max(int(np.ceil(k_frac * n)), 1)
    score = np.maximum(p_hat_A, p_hat_B)
    topk_unsorted = np.argpartition(-score, k - 1)[:k]
    return np.sort(topk_unsorted)


def compute_balanced_class_weights(y) -> tuple[dict, int, int]:
    """Balanced sample-weight scheme equivalent to sklearn's 'balanced':
        w_pos = n / (2 * n_pos),    w_neg = n / (2 * n_neg),    n = n_pos + n_neg.
    Accepts numpy arrays or 1-D torch tensors (CPU/GPU). Returns
    ({0: w_neg, 1: w_pos}, n_pos, n_neg). Asserts both classes are present.
    Weights are NOT clipped — extreme ratios surface as large floats by design.
    HomeCredit's ~3% default rate makes w_pos ≈ 16 and w_neg ≈ 0.5 in practice.
    """
    if isinstance(y, torch.Tensor):
        y_np = y.detach().cpu().numpy()
    else:
        y_np = np.asarray(y)
    y_np = y_np.ravel()
    n_pos = int((y_np > 0.5).sum())
    n_neg = int((y_np <= 0.5).sum())
    n = n_pos + n_neg
    assert n_pos > 0 and n_neg > 0, (
        f"Cannot compute balanced class weights: n_pos={n_pos}, n_neg={n_neg}. "
        "Both classes must be present in the training portion."
    )
    w_pos = n / (2.0 * n_pos)
    w_neg = n / (2.0 * n_neg)
    return {0: float(w_neg), 1: float(w_pos)}, n_pos, n_neg


def make_encoder() -> OrdinalEncoder:
    """OrdinalEncoder configured with the TabReD-style frequency threshold.

    Fit per (window, side) on the full window training rows and reused by every
    bootstrap replica of that side, every Optuna CV fold, and notebooks 03 / 03b.
    """
    return OrdinalEncoder(
        handle_unknown='use_encoded_value',
        unknown_value=CAT_UNK_VALUE,
        min_frequency=CAT_MIN_FREQ,
        encoded_missing_value=CAT_UNK_VALUE,
    )


def make_one_hot() -> OneHotEncoder:
    """OneHotEncoder applied to the OrdinalEncoder's integer output.

    Fit per (window, side) on the same training rows after the OrdinalEncoder,
    shared across XGBoost and LR (notebook 02) and MLP-PLR (this notebook).
    handle_unknown='ignore' makes any code outside the fitted vocabulary (e.g.
    -1 for unseen categories at transform time) become all-zero across that
    cat's dummies — the safe behaviour for nominal one-hot encoding.
    """
    return OneHotEncoder(
        sparse_output=False,
        handle_unknown='ignore',
        dtype=np.float32,
    )


def assemble_combined(X_num_part: np.ndarray,
                      X_cat_str_part: np.ndarray,
                      encoder: OrdinalEncoder,
                      one_hot: OneHotEncoder) -> np.ndarray:
    """Apply pre-fit OrdinalEncoder + OneHotEncoder, hstack with X (numerics + binaries).

    Returns an unscaled `(n, n_num + n_bin + n_oh)` float32 matrix where
    n_oh = sum(len(cats) for cats in one_hot.categories_) is the total dummy
    count for this (window, side). The encoder pair is always pre-fit (per
    (window, side)) — there is no fit-here option. Same shape contract as
    notebook 02's `encode_combined`.
    """
    ord_codes = encoder.transform(X_cat_str_part)
    oh        = one_hot.transform(ord_codes).astype(np.float32)
    return np.hstack([X_num_part, oh])


def scale_combined_inplace(X_combined: np.ndarray,
                           num_scaler: StandardScaler,
                           num_idx) -> np.ndarray:
    """Scale numeric columns; binary and one-hot dummy columns pass through untouched.

    `num_scaler` is per-replica (or per-CV-fold during tuning) and must be pre-fit
    before this call. Dummies are already on the {0, 1} scale and need no scaling.
    """
    out = X_combined.copy()
    out[:, num_idx] = num_scaler.transform(X_combined[:, num_idx])
    return out


def build_arch_config(hparams: dict, n_dummies: int, dummy_col_idx) -> dict:
    """Assemble the dict that reconstructs an MLPPLR from arch-only hyperparameters.

    `n_dummies` and `dummy_col_idx` are window-and-side specific (they depend on
    the fitted OneHotEncoder's vocabulary), so they're passed in rather than
    looked up from globals.
    """
    return {
        'n_num_features': n_num,
        'n_bin_features': n_bin,
        'n_dummies':      int(n_dummies),
        'k_periodic':     int(hparams['k_periodic']),
        'sigma_init':     float(hparams['sigma_init']),
        'd_embedding':    int(hparams['d_embedding']),
        'n_layers':       int(hparams['n_layers']),
        'hidden_dim':     int(hparams['hidden_dim']),
        'dropout':        float(hparams['dropout']),
        'num_col_idx':    [int(i) for i in num_col_idx],
        'bin_col_idx':    [int(i) for i in bin_col_idx],
        'dummy_col_idx':  [int(i) for i in dummy_col_idx],
    }


def train_one_mlp_plr(X_tr_tensor: torch.Tensor,
                      Y_tr_tensor: torch.Tensor,
                      arch_config: dict,
                      train_config: dict,
                      model_seed: int,
                      X_va_tensor: torch.Tensor = None,
                      Y_va_tensor: torch.Tensor = None,
                      class_weights: dict = None):
    """Train one MLP-PLR replica with patience-based early stopping.

    When X_va_tensor / Y_va_tensor are provided (final replica training), those tensors are
    used directly as the early-stopping validation set — they must be the chronological
    tail of the training window, scaled with the per-replica scalers, and extracted
    BEFORE bootstrapping (see main loop).

    When X_va_tensor is None (Optuna tuning phase), the function falls back to an internal
    tail split: the last `val_tail_frac` fraction of X_tr_tensor is held out.

    Class weighting (operational pipeline, always on):
    - If `class_weights` is provided ({0: w_neg, 1: w_pos}), it is used as-is.
      This is how the main loop enforces a single side/window-level mapping
      shared across all R replicas: weights are computed once per side from the
      bootstrap source and reused for every replica.
    - If `class_weights` is None (tuning phase), balanced weights are computed
      from the actual optimization-training portion AFTER any internal tail
      split, so they reflect only the rows the model is fitted on (never the
      internal early-stopping tail).
    The weighted per-row BCE loss is averaged across the batch; the average
    sample weight under this scheme is 1, so the loss scale stays comparable
    to the unweighted setting and does not destabilise the optimiser. With
    HomeCredit's ~3% positive rate, the loss is no longer dominated by the
    correctly-classified majority — gradients on positives are amplified by
    w_pos ≈ 16 while negatives are scaled down by w_neg ≈ 0.5.
    Validation PR-AUC and the early-stopping signal remain unweighted.

    Returns: (best_state_dict_on_cpu, training_log_df, best_epoch, best_val_pr_auc)
    """
    torch.manual_seed(model_seed)
    torch.cuda.manual_seed_all(model_seed)
    np.random.seed(model_seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark     = False

    if X_va_tensor is None:
        n     = X_tr_tensor.shape[0]
        n_val = max(int(np.ceil(n * train_config['val_tail_frac'])), 1)
        X_tr  = X_tr_tensor[:-n_val]
        Y_tr  = Y_tr_tensor[:-n_val]
        X_va  = X_tr_tensor[-n_val:]
        Y_va  = Y_tr_tensor[-n_val:]
    else:
        X_tr, Y_tr = X_tr_tensor, Y_tr_tensor
        X_va, Y_va = X_va_tensor, Y_va_tensor

    # ── Resolve class weights for the actual optimization-training portion Y_tr ──
    if class_weights is None:
        class_weights, _, _ = compute_balanced_class_weights(Y_tr)
    w_pos = float(class_weights[1])
    w_neg = float(class_weights[0])
    sw_tr = torch.where(Y_tr > 0.5,
                        torch.tensor(w_pos, device=Y_tr.device, dtype=torch.float32),
                        torch.tensor(w_neg, device=Y_tr.device, dtype=torch.float32))

    model = MLPPLR(**arch_config).to(DEVICE)
    opt   = torch.optim.AdamW(
        model.parameters(),
        lr=train_config['learning_rate'],
        weight_decay=train_config['weight_decay'],
    )
    # Per-row reduction so we can multiply by sample weights before averaging.
    loss_fn = nn.BCEWithLogitsLoss(reduction='none')

    best_val_pr   = -1.0
    best_state    = None
    best_epoch    = -1
    patience_left = train_config['patience']
    bs            = train_config['batch_size']
    n_tr          = X_tr.shape[0]

    log_rows = []

    for epoch in range(1, train_config['max_epochs'] + 1):
        model.train()
        perm = torch.randperm(n_tr, device=DEVICE)
        total_loss = 0.0
        for start in range(0, n_tr, bs):
            sel   = perm[start:start + bs]
            xb    = X_tr[sel]
            yb    = Y_tr[sel]
            wb    = sw_tr[sel]
            opt.zero_grad()
            logit = model(xb)
            # Mean of weighted per-row losses. Average sample weight under the
            # balanced scheme equals 1, so the loss scale stays comparable to
            # unweighted training and Adam's effective step size is preserved.
            loss  = (loss_fn(logit, yb) * wb).mean()
            loss.backward()
            opt.step()
            total_loss += loss.item() * xb.shape[0]
        avg_loss = total_loss / n_tr

        # Validation PR-AUC — unweighted, mirrors the rest of the pipeline.
        model.eval()
        val_probs_chunks = []
        with torch.no_grad():
            for start in range(0, X_va.shape[0], 2048):
                logit = model(X_va[start:start + 2048])
                val_probs_chunks.append(torch.sigmoid(logit).cpu().numpy())
        val_probs = np.concatenate(val_probs_chunks)
        val_y     = Y_va.cpu().numpy()
        val_pr    = float(average_precision_score(val_y, val_probs))

        log_rows.append({'epoch': epoch, 'train_loss': avg_loss, 'val_pr_auc': val_pr})

        if val_pr > best_val_pr:
            best_val_pr   = val_pr
            best_state    = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            best_epoch    = epoch
            patience_left = train_config['patience']
        else:
            patience_left -= 1
            if patience_left <= 0:
                break

    if best_state is None:
        best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        best_epoch = len(log_rows)

    del model
    torch.cuda.empty_cache()

    return best_state, pd.DataFrame(log_rows), best_epoch, best_val_pr


def predict_proba_mlp_plr(model: nn.Module, X_tensor: torch.Tensor,
                          batch_size: int = 2048) -> np.ndarray:
    """Forward pass in eval mode, apply sigmoid, return 1-D numpy float32."""
    model.eval()
    out = []
    with torch.no_grad():
        for start in range(0, X_tensor.shape[0], batch_size):
            logit = model(X_tensor[start:start + batch_size])
            out.append(torch.sigmoid(logit).cpu().numpy())
    return np.concatenate(out).astype(np.float32)


def tune_hyperparameters_mlp_plr(window_idx: np.ndarray,
                                  X_all: np.ndarray,
                                  X_cat_all: np.ndarray,
                                  Y_all: np.ndarray,
                                  encoder: OrdinalEncoder,
                                  one_hot: OneHotEncoder,
                                  study_seed: int,
                                  n_trials: int = N_TRIALS_MLP) -> dict:
    """Optuna MLP-PLR tuning on the given training window.

    Both `encoder` and `one_hot` are the shared per-(window, side) artefacts pre-fit
    in the main loop; they are reused across all CV folds. Per CV fold, only a fresh
    numeric `StandardScaler` is fit on the fold's training portion.

    Class weighting: train_one_mlp_plr is called without an explicit class_weights
    dict, so balanced weights are computed inside the trainer from the actual
    optimization-training portion (post internal ES-tail split). The fold-validation
    portion is never used to derive weights, and the PR-AUC scored on it is unweighted.
    """
    X_win     = X_all[window_idx]      # preserves chronological step-sorted order
    X_cat_win = X_cat_all[window_idx]
    Y_win     = Y_all[window_idx]

    # Encode + one-hot once for the whole window — both transforms are shared.
    X_win_combined_unscaled = assemble_combined(X_win, X_cat_win, encoder, one_hot)
    n_dummies     = X_win_combined_unscaled.shape[1] - n_num_bin
    dummy_col_idx = list(range(n_num_bin, X_win_combined_unscaled.shape[1]))

    tscv = TimeSeriesSplit(n_splits=CV_N_SPLITS)

    def objective(trial: optuna.Trial) -> float:
        hp = {
            'k_periodic':    trial.suggest_int('k_periodic', 4, 16),
            'sigma_init':    trial.suggest_float('sigma_init', 0.01, 10.0, log=True),
            'd_embedding':   trial.suggest_categorical('d_embedding', [8, 16, 24]),
            'n_layers':      trial.suggest_int('n_layers', 1, 3),
            'hidden_dim':    trial.suggest_categorical('hidden_dim', [128, 256, 512]),
            'dropout':       trial.suggest_float('dropout', 0.0, 0.5),
            'learning_rate': trial.suggest_float('learning_rate', 1e-4, 1e-2, log=True),
            'weight_decay':  trial.suggest_float('weight_decay',  1e-6, 1e-2, log=True),
        }

        arch_config  = build_arch_config(hp, n_dummies, dummy_col_idx)
        train_config = {
            'learning_rate': hp['learning_rate'],
            'weight_decay':  hp['weight_decay'],
            'max_epochs':    MLP_FIXED['tuning_max_epochs'],
            'patience':      MLP_FIXED['patience'],
            'batch_size':    MLP_FIXED['batch_size'],
            'val_tail_frac': MLP_FIXED['val_tail_frac'],
        }

        pr_aucs = []
        for fold, (tr, va) in enumerate(tscv.split(X_win_combined_unscaled)):
            # Per-fold numeric scaler on fold-train only.
            num_scaler = StandardScaler().fit(X_win_combined_unscaled[tr][:, num_col_idx])

            X_tr_s = scale_combined_inplace(X_win_combined_unscaled[tr], num_scaler, num_col_idx)
            X_va_s = scale_combined_inplace(X_win_combined_unscaled[va], num_scaler, num_col_idx)

            X_tr_t = torch.from_numpy(X_tr_s).float().to(DEVICE)
            Y_tr_t = torch.from_numpy(Y_win[tr].astype(np.float32)).to(DEVICE)
            X_va_t = torch.from_numpy(X_va_s).float().to(DEVICE)

            # class_weights=None → trainer computes balanced weights from the post-split
            # optimization-training portion only (excludes the internal ES tail).
            best_state, _, _, _ = train_one_mlp_plr(
                X_tr_t, Y_tr_t, arch_config, train_config,
                model_seed=study_seed + fold,
            )

            pred_model = MLPPLR(**arch_config).to(DEVICE)
            pred_model.load_state_dict({k: v.to(DEVICE) for k, v in best_state.items()})
            probs = predict_proba_mlp_plr(pred_model, X_va_t)
            pr_aucs.append(float(average_precision_score(Y_win[va], probs)))

            del pred_model, X_tr_t, Y_tr_t, X_va_t
            torch.cuda.empty_cache()

        return float(np.mean(pr_aucs))

    sampler = TPESampler(seed=study_seed)
    study   = optuna.create_study(direction='maximize', sampler=sampler)
    study.optimize(objective, n_trials=n_trials, show_progress_bar=False)
    return dict(study.best_trial.params)


def assert_binary_slice(y, name):
    assert len(np.unique(y)) == 2, f'{name} contains only one class.'


print('Helpers defined.')
````

## Main training loop — one iteration per window pair

````python
SEED_BASE = 42   # Shared with notebook 02 — do not change.

# ── Class-weighting policy (operational retraining pipeline) ────────────────
# Balanced sample weights (sklearn-equivalent: w_c = n / (2 * n_c)) are applied
# to every fit — Optuna folds and replica fits. Weights are computed from the
# actual optimization-training portion only:
#  - tuning: post internal ES-tail split inside train_one_mlp_plr;
#  - replica: from the bootstrap source (training core, before the external
#    chronological ES tail). All R replicas of a side share one mapping.
# Validation/evaluation PR-AUC and the early-stopping signal remain unweighted.
# HomeCredit's ~3% default rate makes w_pos ≈ 16 and w_neg ≈ 0.5 — the asymmetry
# is by design and is NOT clipped.
CLASS_WEIGHTING = {
    'enabled':                    True,
    'scheme':                     'balanced_sample_weights',
    'formula':                    'w_pos = n / (2 * n_pos); w_neg = n / (2 * n_neg)',
    'clipped':                    False,
    'evaluation_pr_auc_weighted': False,
    'replica_data_scope':         'training_core_before_early_stopping_tail',
    'tuning_data_scope':          'fold_training_post_internal_es_split',
    'loss_normalization':         'mean_of_weighted_per_row_BCE',
}

CURRENT_RUN_PARAMS = {
    'model_type':  MODEL_TYPE,
    'seed_base':   SEED_BASE,

    'window_parameters': {
        'L':           int(config['parameters']['L']),
        'S':           int(config['parameters']['S']),
        'H':           int(config['parameters']['H']),
        'R':           int(R),
        'K_FRAC':      float(K_FRAC),
        'PAIR_STRIDE': int(config['parameters'].get('PAIR_STRIDE', 1)),
    },

    'cv_n_splits':      int(CV_N_SPLITS),
    'n_trials_mlp':     int(N_TRIALS_MLP),
    'mlp_fixed':        MLP_FIXED,
    'cat_min_freq':     float(CAT_MIN_FREQ),
    'cat_unk_value':    int(CAT_UNK_VALUE),
    'encoder_scope':    'per_window_side_shared',
    'cat_encoding':     'one_hot_per_window_l2_threshold',
    'bundle_schema':    'mlp_plr_bundle_v3_homecredit_oh',

    # Class weighting (scheme metadata; per-pair n_pos/n_neg/w_pos/w_neg are
    # written to class_weights.json inside each pair directory).
    'class_weighting':  CLASS_WEIGHTING,
}

performance_log = []

for p in pairs:
    pid                = p['pair_id']
    pair_dir           = MODEL_DIR / f'pair_{pid:02d}'
    pred_file          = pair_dir / 'predictions.npz'
    run_params_file    = pair_dir / 'run_params.json'
    class_weights_file = pair_dir / 'class_weights.json'

    # ── Skip-if-done: validate full run configuration against saved run_params ──
    # Checking only for predictions.npz existence is insufficient: if tuning,
    # model, window, selection, or class-weighting settings change, existing
    # predictions are stale.
    if pred_file.exists() and run_params_file.exists():
        with open(run_params_file) as f:
            saved = json.load(f)
        params_match = (saved == CURRENT_RUN_PARAMS)

        if not class_weights_file.exists():
            print(f'Pair {pid:02d}: class_weights.json missing — re-running.')
        elif params_match:
            data = np.load(pred_file)
            if all(k in data.files for k in ('roc_auc_A', 'roc_auc_B')):
                print(f'Pair {pid:02d}: already done, skipping.')
                performance_log.append({
                    'pair_id':   pid,
                    'pr_auc_A':  float(data['pr_auc_A']),
                    'pr_auc_B':  float(data['pr_auc_B']),
                    'roc_auc_A': float(data['roc_auc_A']),
                    'roc_auc_B': float(data['roc_auc_B']),
                    'n_flagged': int(data['flagged_idx'].shape[0]),
                })
                continue
            else:
                print(f'Pair {pid:02d}: stale predictions.npz (missing roc_auc keys) — re-running.')
        else:
            changed_keys = [k for k in CURRENT_RUN_PARAMS if saved.get(k) != CURRENT_RUN_PARAMS[k]]
            print(f'Pair {pid:02d}: run_params changed in {changed_keys} — re-running.')
    elif pred_file.exists():
        print(f'Pair {pid:02d}: run_params.json missing — re-running.')

    print(f'\n── [mlp_plr] Pair {pid:02d}: A_end={p["step_label_A"]}  B_end={p["step_label_B"]}  '
          f'eval={p["eval_start_label"]}→{p["eval_end_label"]}  '
          f'|A|={p["n_train_A"]:,} |B|={p["n_train_B"]:,} |eval|={p["n_eval"]:,} ──')

    idx_A    = np.array(p['idx_A'],    dtype=np.int64)
    idx_B    = np.array(p['idx_B'],    dtype=np.int64)
    idx_eval = np.array(p['idx_eval'], dtype=np.int64)

    assert len(set(idx_A.tolist()) & set(idx_eval.tolist())) == 0, 'idx_A overlaps idx_eval!'
    assert len(set(idx_B.tolist()) & set(idx_eval.tolist())) == 0, 'idx_B overlaps idx_eval!'

    X_eval     = X[idx_eval]
    X_cat_eval = X_cat[idx_eval]
    Y_eval     = Y[idx_eval]
    assert_binary_slice(Y[idx_A], f'pair {pid:02d} window A')
    assert_binary_slice(Y[idx_B], f'pair {pid:02d} window B')
    assert_binary_slice(Y_eval,   f'pair {pid:02d} eval slice')

    pair_dir.mkdir(parents=True, exist_ok=True)
    dir_A = pair_dir / 'replicas_A'; dir_A.mkdir(exist_ok=True)
    dir_B = pair_dir / 'replicas_B'; dir_B.mkdir(exist_ok=True)

    # ── Reference scaler (numerics on window A) — consumed by nb 04 for cross-pair drift ──
    ref_scaler = StandardScaler().fit(X[idx_A][:, num_col_idx])
    joblib.dump(ref_scaler, pair_dir / 'reference_scaler.joblib')

    # ── Shared per-(window, side) OrdinalEncoders (used by replicas, tuning, and 03 / 03b) ──
    encoder_A = make_encoder()
    encoder_A.fit(X_cat[idx_A])
    joblib.dump(encoder_A, dir_A / 'encoder.joblib')

    encoder_B = make_encoder()
    encoder_B.fit(X_cat[idx_B])
    joblib.dump(encoder_B, dir_B / 'encoder.joblib')

    # ── Shared per-(window, side) OneHotEncoders (used by every model; same artefact as nb 02) ──
    # Fit on the OrdinalEncoder's integer output from the full window training rows;
    # reused by every replica of that side and across all Optuna CV folds.
    one_hot_A = make_one_hot()
    one_hot_A.fit(encoder_A.transform(X_cat[idx_A]))
    joblib.dump(one_hot_A, dir_A / 'one_hot.joblib')

    one_hot_B = make_one_hot()
    one_hot_B.fit(encoder_B.transform(X_cat[idx_B]))
    joblib.dump(one_hot_B, dir_B / 'one_hot.joblib')

    n_oh_A = sum(len(c) for c in one_hot_A.categories_)
    n_oh_B = sum(len(c) for c in one_hot_B.categories_)
    dummy_col_idx_A = list(range(n_num_bin, n_num_bin + n_oh_A))
    dummy_col_idx_B = list(range(n_num_bin, n_num_bin + n_oh_B))
    print(f'  one-hot dummies: A={n_oh_A}, B={n_oh_B}  (n_cat={n_cat} cat variables)')

    # ── Tune hyperparameters per window (shared encoder + one_hot) ──────────
    print('  Tuning window A ...')
    hparams_A = tune_hyperparameters_mlp_plr(
        idx_A, X, X_cat, Y, encoder_A, one_hot_A,
        study_seed=SEED_BASE + pid * 10 + 1, n_trials=N_TRIALS_MLP,
    )
    with open(pair_dir / 'hparams_A.json', 'w') as f:
        json.dump(hparams_A, f, indent=2)
    print(f'    best A params: {hparams_A}')

    print('  Tuning window B ...')
    hparams_B = tune_hyperparameters_mlp_plr(
        idx_B, X, X_cat, Y, encoder_B, one_hot_B,
        study_seed=SEED_BASE + pid * 10 + 2, n_trials=N_TRIALS_MLP,
    )
    with open(pair_dir / 'hparams_B.json', 'w') as f:
        json.dump(hparams_B, f, indent=2)
    print(f'    best B params: {hparams_B}')

    # ── Chronological ES validation split (done before bootstrap, shared across replicas) ──
    val_frac = MLP_FIXED['val_tail_frac']

    n_es_A           = max(int(np.ceil(len(idx_A) * val_frac)), 1)
    idx_A_boot       = idx_A[:-n_es_A]                # bootstrap source (excludes chronological tail)
    X_va_A_raw       = X[idx_A[-n_es_A:]]             # raw ES validation features (scaled per-replica)
    X_cat_va_A_raw   = X_cat[idx_A[-n_es_A:]]
    Y_va_A           = Y[idx_A[-n_es_A:]].astype(np.float32)

    n_es_B           = max(int(np.ceil(len(idx_B) * val_frac)), 1)
    idx_B_boot       = idx_B[:-n_es_B]
    X_va_B_raw       = X[idx_B[-n_es_B:]]
    X_cat_va_B_raw   = X_cat[idx_B[-n_es_B:]]
    Y_va_B           = Y[idx_B[-n_es_B:]].astype(np.float32)

    assert_binary_slice(Y_va_A,        f'pair {pid:02d} window A early-stopping tail')
    assert_binary_slice(Y_va_B,        f'pair {pid:02d} window B early-stopping tail')
    assert_binary_slice(Y[idx_A_boot], f'pair {pid:02d} window A bootstrap source')
    assert_binary_slice(Y[idx_B_boot], f'pair {pid:02d} window B bootstrap source')

    # ── Side/window-level balanced class weights (computed once per side) ──────
    # Computed from the bootstrap source (training core, excluding the chronological
    # ES tail) so they describe the actual optimization-training pool.  All R
    # replicas of a side reuse the same {w_pos, w_neg} mapping; only the bootstrap
    # sample and the model RNG vary between replicas. A and B may end up with
    # different weights because their class balance can differ across temporal
    # windows; this is intentional and is part of what we measure as end-to-end
    # explanation drift.
    cw_A_side, n_pos_A, n_neg_A = compute_balanced_class_weights(Y[idx_A_boot])
    cw_B_side, n_pos_B, n_neg_B = compute_balanced_class_weights(Y[idx_B_boot])
    print(f'  Class weights A (training_core_before_early_stopping_tail): '
          f'n_pos={n_pos_A:,}  n_neg={n_neg_A:,}  '
          f'w_pos={cw_A_side[1]:.4f}  w_neg={cw_A_side[0]:.4f}')
    print(f'  Class weights B (training_core_before_early_stopping_tail): '
          f'n_pos={n_pos_B:,}  n_neg={n_neg_B:,}  '
          f'w_pos={cw_B_side[1]:.4f}  w_neg={cw_B_side[0]:.4f}')

    pair_class_weights_meta = {
        'scheme':                     'balanced_sample_weights',
        'formula':                    'w_pos = n / (2 * n_pos); w_neg = n / (2 * n_neg)',
        'clipped':                    False,
        'evaluation_pr_auc_weighted': False,
        'replica_data_scope':         'training_core_before_early_stopping_tail',
        'A': {
            'n_pos': int(n_pos_A), 'n_neg': int(n_neg_A),
            'w_pos': float(cw_A_side[1]), 'w_neg': float(cw_A_side[0]),
            'data_scope': 'training_core_before_early_stopping_tail',
        },
        'B': {
            'n_pos': int(n_pos_B), 'n_neg': int(n_neg_B),
            'w_pos': float(cw_B_side[1]), 'w_neg': float(cw_B_side[0]),
            'data_scope': 'training_core_before_early_stopping_tail',
        },
    }

    # ── Train R replicas for window A ──────────────────────────────────────
    preds_A          = np.zeros((R, len(idx_eval)), dtype=np.float32)
    per_rep_pr_auc_A = np.zeros(R, dtype=np.float32)

    for r in range(R):
        boot_seed  = SEED_BASE + pid * 10_000 + r * 2          # A: even offset
        model_seed = SEED_BASE + pid * 10_000 + r * 2 + 1      # A: odd offset

        boot_idx = stratified_bootstrap(idx_A_boot, Y, seed=boot_seed)

        # Encode + one-hot using shared encoder_A + one_hot_A (no fit here).
        X_tr_combined_unscaled   = assemble_combined(X[boot_idx], X_cat[boot_idx], encoder_A, one_hot_A)
        X_va_combined_unscaled   = assemble_combined(X_va_A_raw,  X_cat_va_A_raw,  encoder_A, one_hot_A)
        X_eval_combined_unscaled = assemble_combined(X_eval,      X_cat_eval,      encoder_A, one_hot_A)

        # Per-replica numeric scaler on this bootstrap's numerics; dummies pass through raw 0/1.
        num_scaler = StandardScaler().fit(X_tr_combined_unscaled[:, num_col_idx])

        X_tr_s   = scale_combined_inplace(X_tr_combined_unscaled,   num_scaler, num_col_idx)
        X_va_s   = scale_combined_inplace(X_va_combined_unscaled,   num_scaler, num_col_idx)
        X_eval_s = scale_combined_inplace(X_eval_combined_unscaled, num_scaler, num_col_idx)

        X_tr_t   = torch.from_numpy(X_tr_s).float().to(DEVICE)
        Y_tr_t   = torch.from_numpy(Y[boot_idx].astype(np.float32)).to(DEVICE)
        X_va_t   = torch.from_numpy(X_va_s).float().to(DEVICE)
        Y_va_t   = torch.from_numpy(Y_va_A).to(DEVICE)
        X_eval_t = torch.from_numpy(X_eval_s).float().to(DEVICE)

        arch_config  = build_arch_config(hparams_A, n_oh_A, dummy_col_idx_A)
        train_config = {
            'learning_rate': hparams_A['learning_rate'],
            'weight_decay':  hparams_A['weight_decay'],
            'max_epochs':    MLP_FIXED['max_epochs'],
            'patience':      MLP_FIXED['patience'],
            'batch_size':    MLP_FIXED['batch_size'],
            'val_tail_frac': val_frac,
        }

        # Pass the side-level class-weight mapping explicitly so the trainer
        # uses it as-is rather than recomputing per replica.
        best_state, training_log, best_epoch, best_val_pr = train_one_mlp_plr(
            X_tr_t, Y_tr_t, arch_config, train_config, model_seed=model_seed,
            X_va_tensor=X_va_t, Y_va_tensor=Y_va_t,
            class_weights=cw_A_side,
        )
        training_log.to_csv(dir_A / f'training_log_r{r}.csv', index=False)

        # Reconstruct best model for prediction.
        pred_model = MLPPLR(**arch_config).to(DEVICE)
        pred_model.load_state_dict({k: v.to(DEVICE) for k, v in best_state.items()})
        preds_A[r] = predict_proba_mlp_plr(pred_model, X_eval_t)
        per_rep_pr_auc_A[r] = float(average_precision_score(Y_eval, preds_A[r]))

        # Per-replica bundle: state_dict, arch_config, and the per-replica numeric
        # StandardScaler. The shared encoder + one_hot live in replicas_A/.
        bundle = {
            'state_dict':  best_state,
            'arch_config': arch_config,
            'scaler':      num_scaler,
        }
        joblib.dump(bundle, dir_A / f'model_r{r}.joblib', compress=3)

        with open(dir_A / f'seeds_r{r}.json', 'w') as f:
            json.dump({'bootstrap_seed': int(boot_seed),
                       'model_seed':     int(model_seed)}, f, indent=2)

        print(f'  A replica {r}: PR-AUC = {per_rep_pr_auc_A[r]:.4f}  '
              f'(best epoch {best_epoch}, val PR-AUC {best_val_pr:.4f})')

        del pred_model, X_tr_t, Y_tr_t, X_va_t, Y_va_t, X_eval_t
        torch.cuda.empty_cache()

    # ── Train R replicas for window B ──────────────────────────────────────
    preds_B          = np.zeros((R, len(idx_eval)), dtype=np.float32)
    per_rep_pr_auc_B = np.zeros(R, dtype=np.float32)

    for r in range(R):
        boot_seed  = SEED_BASE + pid * 10_000 + 5_000 + r * 2
        model_seed = SEED_BASE + pid * 10_000 + 5_000 + r * 2 + 1

        boot_idx = stratified_bootstrap(idx_B_boot, Y, seed=boot_seed)

        X_tr_combined_unscaled   = assemble_combined(X[boot_idx], X_cat[boot_idx], encoder_B, one_hot_B)
        X_va_combined_unscaled   = assemble_combined(X_va_B_raw,  X_cat_va_B_raw,  encoder_B, one_hot_B)
        X_eval_combined_unscaled = assemble_combined(X_eval,      X_cat_eval,      encoder_B, one_hot_B)

        num_scaler = StandardScaler().fit(X_tr_combined_unscaled[:, num_col_idx])

        X_tr_s   = scale_combined_inplace(X_tr_combined_unscaled,   num_scaler, num_col_idx)
        X_va_s   = scale_combined_inplace(X_va_combined_unscaled,   num_scaler, num_col_idx)
        X_eval_s = scale_combined_inplace(X_eval_combined_unscaled, num_scaler, num_col_idx)

        X_tr_t   = torch.from_numpy(X_tr_s).float().to(DEVICE)
        Y_tr_t   = torch.from_numpy(Y[boot_idx].astype(np.float32)).to(DEVICE)
        X_va_t   = torch.from_numpy(X_va_s).float().to(DEVICE)
        Y_va_t   = torch.from_numpy(Y_va_B).to(DEVICE)
        X_eval_t = torch.from_numpy(X_eval_s).float().to(DEVICE)

        arch_config  = build_arch_config(hparams_B, n_oh_B, dummy_col_idx_B)
        train_config = {
            'learning_rate': hparams_B['learning_rate'],
            'weight_decay':  hparams_B['weight_decay'],
            'max_epochs':    MLP_FIXED['max_epochs'],
            'patience':      MLP_FIXED['patience'],
            'batch_size':    MLP_FIXED['batch_size'],
            'val_tail_frac': val_frac,
        }

        best_state, training_log, best_epoch, best_val_pr = train_one_mlp_plr(
            X_tr_t, Y_tr_t, arch_config, train_config, model_seed=model_seed,
            X_va_tensor=X_va_t, Y_va_tensor=Y_va_t,
            class_weights=cw_B_side,
        )
        training_log.to_csv(dir_B / f'training_log_r{r}.csv', index=False)

        pred_model = MLPPLR(**arch_config).to(DEVICE)
        pred_model.load_state_dict({k: v.to(DEVICE) for k, v in best_state.items()})
        preds_B[r] = predict_proba_mlp_plr(pred_model, X_eval_t)
        per_rep_pr_auc_B[r] = float(average_precision_score(Y_eval, preds_B[r]))

        bundle = {
            'state_dict':  best_state,
            'arch_config': arch_config,
            'scaler':      num_scaler,
        }
        joblib.dump(bundle, dir_B / f'model_r{r}.joblib', compress=3)

        with open(dir_B / f'seeds_r{r}.json', 'w') as f:
            json.dump({'bootstrap_seed': int(boot_seed),
                       'model_seed':     int(model_seed)}, f, indent=2)

        print(f'  B replica {r}: PR-AUC = {per_rep_pr_auc_B[r]:.4f}  '
              f'(best epoch {best_epoch}, val PR-AUC {best_val_pr:.4f})')

        del pred_model, X_tr_t, Y_tr_t, X_va_t, Y_va_t, X_eval_t
        torch.cuda.empty_cache()

    # ── Replica-averaged predictions and flagged set ──────────────────────
    # Predictions are ordinary positive-class probabilities — sample weights affect
    # the training objective only, not the model's output transformation.
    p_hat_A = preds_A.mean(axis=0)
    p_hat_B = preds_B.mean(axis=0)
    flagged_local_idx = compute_flagged_topk(p_hat_A, p_hat_B, K_FRAC)

    # Validation/evaluation PR-AUC and ROC-AUC are unweighted by construction.
    pr_auc_A  = float(average_precision_score(Y_eval, p_hat_A))
    pr_auc_B  = float(average_precision_score(Y_eval, p_hat_B))
    roc_auc_A = float(roc_auc_score(Y_eval, p_hat_A))
    roc_auc_B = float(roc_auc_score(Y_eval, p_hat_B))

    print(f'  Averaged: PR-AUC A = {pr_auc_A:.4f}, PR-AUC B = {pr_auc_B:.4f}')
    print(f'            ROC-AUC A = {roc_auc_A:.4f}, ROC-AUC B = {roc_auc_B:.4f}')
    print(f'  Flagged (top {K_FRAC:.0%}): {flagged_local_idx.shape[0]:,} / {len(idx_eval):,}')

    # ── Save predictions NPZ (schema identical to nb 02) ─────────────────
    np.savez_compressed(
        pred_file,
        preds_A              = preds_A,
        preds_B              = preds_B,
        p_hat_A              = p_hat_A.astype(np.float32),
        p_hat_B              = p_hat_B.astype(np.float32),
        flagged_idx          = flagged_local_idx,
        Y_eval               = Y_eval,
        pr_auc_A             = np.float32(pr_auc_A),
        pr_auc_B             = np.float32(pr_auc_B),
        roc_auc_A            = np.float32(roc_auc_A),
        roc_auc_B            = np.float32(roc_auc_B),
        per_replica_pr_auc_A = per_rep_pr_auc_A,
        per_replica_pr_auc_B = per_rep_pr_auc_B,
    )

    with open(run_params_file, 'w') as f:
        json.dump(CURRENT_RUN_PARAMS, f, indent=2)

    # ── Save per-pair class-weighting metadata for transparency ──────────
    # The scheme itself is part of the staleness fingerprint via CURRENT_RUN_PARAMS;
    # this file records the realised n_pos / n_neg / w_pos / w_neg per side.
    with open(class_weights_file, 'w') as f:
        json.dump(pair_class_weights_meta, f, indent=2)

    performance_log.append({
        'pair_id':   pid,
        'pr_auc_A':  pr_auc_A,
        'pr_auc_B':  pr_auc_B,
        'roc_auc_A': roc_auc_A,
        'roc_auc_B': roc_auc_B,
        'n_flagged': flagged_local_idx.shape[0],
    })

print('\n✓ All window pairs processed.')
````
